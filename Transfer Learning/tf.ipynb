The input for VGG16 is originally (224, 224), changed it to (64, 64) so i could run this code locally on my pc
the dataset used it rally small with only 72 files belonging to two classes, which is really small for the VGG model so the model is expected to underfit
this noteboook is used for learning purpose strictly 

```python
from tensorflow.keras.applications import VGG16 
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import image_dataset_from_directory

train_ds = image_dataset_from_directory(
    directory = '/home/twelpoundmuffin/Desktop/Required Datasets/dl/cat and dog',
    labels = 'inferred',
    batch_size = 4, 
    label_mode = 'int',
    image_size = (64, 64)

)

model1 = VGG16(weights = 'imagenet', include_top = False, input_shape = (64, 64, 3))
for layer in model1.layers:
    layer.trainable = False

model = Sequential([
    model1,
    Flatten(),
    Dense(64, activation='relu'), 
    Dense(2, activation='softmax')
])
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])

model.fit(train_ds, epochs = 10)
```
